### Step 6. Count how many unique items are in the dataset - Method 1

**Explanation:**  
`.nunique()` counts the number of unique (distinct) values in a column.

**Difference from `.unique()`:**
- `.unique()` returns an **array of the actual values**
- `.nunique()` returns a **single number** (the count)

**Example:**
- If a column has values: ['A', 'B', 'A', 'C', 'B']
- `.unique()` returns: ['A', 'B', 'C']
- `.nunique()` returns: 3

````python
chipo['item_name'].nunique()
````

**Expected Output:**
```
50
```

This means there are **50 different menu items** in the Chipotle dataset.

### Step 6 (Alternative). Count unique items - Method 2

**Explanation:**  
This is an **alternative approach** to count unique items by combining two methods:

1. **`.unique()`**: Gets the array of unique values
2. **`len()`**: Counts how many elements are in that array

**Comparison:**
- **Method 1** (above): `chipo['item_name'].nunique()` → Direct and efficient
- **Method 2** (this): `len(chipo['item_name'].unique())` → Two-step process

**Which is better?**  
`.nunique()` is **preferred** because:
- More readable and explicit
- Slightly more efficient (doesn't need to store the array in memory)
- Standard pandas convention

**When to use `len(.unique())`:**  
When you also need the actual unique values for other operations.

````python
len(chipo['item_name'].unique())
````

**Expected Output:**
```
50
```

Same result as Method 1, confirming there are 50 unique items.

### Step 7. Convert item_price from string to float

**Explanation:**  
Currently, `item_price` is stored as a string (object type) because it includes the '$' symbol. We need to convert it to a numeric type to perform calculations.

**Breaking down the code:**

1. **`chipo['item_price'].apply(lambda x: x[1:])`**:
   - `.apply()` applies a function to each value in the column
   - `lambda x: x[1:]` is an anonymous function that:
     - Takes each value `x` (e.g., '$2.39')
     - Slices from index 1 onwards: `x[1:]` removes the first character '$'
     - Returns '2.39' (as a string)

2. **`.astype(float)`**:
   - Converts the string '2.39' to the float 2.39
   - Now we can perform mathematical operations

3. **Assignment**: Replaces the original column with the numeric version

**Why is this important?**  
Without this conversion, operations like `.max()`, `.sum()`, or `.mean()` won't work properly on price data.

````python
chipo['item_price'] = chipo['item_price'].apply(lambda x: x[1:]).astype(float)
````

**Verification:**  
Let's check the data type changed:

````python
chipo.info()
````

**Note:** Now `item_price` shows as `float64` instead of `object`

### Step 8. Create a new column for unit price

**Explanation:**  
We're creating a new column `item_price_unit` that calculates the **price per individual item**.

**Formula:**  
`unit_price = total_price / quantity`

**Example:**
- If someone orders 2 burritos for $16.98
- `item_price` = 16.98 (total)
- `quantity` = 2
- `item_price_unit` = 16.98 / 2 = 8.49 (price per burrito)

**Key concept:**  
This is **vectorized operation** - pandas automatically divides each row's price by its quantity without needing a loop.

**Why is this useful?**  
Allows us to compare individual item prices across orders with different quantities.

````python
chipo['item_price_unit'] = chipo['item_price'] / chipo['quantity']
chipo.head()
````

### Step 10. Find the most expensive item - Method 1 (using filtering)

**Explanation:**  
This approach uses **boolean indexing** to find the row with the maximum price.

**Breaking down the code:**

1. **`chipo['item_price'].max()`**:
   - Finds the maximum value in the `item_price` column
   - Returns a single number (e.g., 44.25)

2. **`chipo['item_price'] == chipo['item_price'].max()`**:
   - Creates a **boolean mask** (True/False for each row)
   - True where the price equals the maximum
   - False everywhere else

3. **`chipo[...]`**:
   - Filters the DataFrame using the boolean mask
   - Returns only rows where the condition is True

**Result:** The complete row(s) with the highest price

**Note:** If multiple rows have the same maximum price, all will be returned.

````python
chipo[chipo['item_price'] == chipo['item_price'].max()]
````

**Expected Output:**
```
          quantity                     item_name choice_description  item_price  \
order_id                                                                          
1443            15  Chips and Fresh Tomato Salsa                NaN       44.25   

          item_price_unit  
order_id                   
1443                 2.95  
```

**Analysis:**  
The most expensive single order was 15 orders of Chips and Fresh Tomato Salsa for a total of $44.25 ($2.95 each).

### Step 10 (Alternative). Find most expensive item - Method 2 (using idxmax)

**Explanation:**  
`.idxmax()` returns the **index label** of the row with the maximum value.

**Breaking down the code:**

1. **`chipo['item_price'].idxmax()`**:
   - Finds which index has the maximum price
   - Returns the index label (in this case, order_id = 1443)
   - Does NOT return the price itself, just the index

2. **`.loc[...]`**:
   - Uses the index label to retrieve the entire row

**Comparison with Method 1:**
- **Method 1**: `chipo[chipo['item_price'] == chipo['item_price'].max()]`
  - More intuitive for beginners
  - Returns multiple rows if there are ties
  
- **Method 2**: `chipo.loc[chipo['item_price'].idxmax()]`
  - More efficient (only one comparison)
  - Returns only the first row if there are ties
  - Preferred in professional code

**Related methods:**
- `.idxmin()` for finding the index of the minimum value
- `.idxmax()` and `.idxmin()` are useful when you need the location, not just the value

````python
chipo.loc[chipo['item_price'].idxmax()]
````

**Expected Output:**
```
quantity                                     15
item_name             Chips and Fresh Tomato Salsa
choice_description                          NaN
item_price                                44.25
item_price_unit                            2.95
Name: 1443, dtype: object
```

Same result as Method 1, but displayed as a Series instead of a DataFrame row.

## Advanced Filtering and Counting

### Step 11. What was the quantity of the most expensive item ordered?

**Approach 1: Chain filtering and value extraction**

**Explanation:**  
This is a **three-step process** chained together:

1. **`chipo[chipo['item_price'] == chipo['item_price'].max()]`**:
   - Filters to get the row(s) with the maximum price
   - Returns a DataFrame (even if it's just one row)

2. **`['quantity']`**:
   - Selects just the 'quantity' column from the filtered DataFrame
   - Returns a Series

3. **`.values[0]`**:
   - `.values` converts the Series to a numpy array
   - `[0]` gets the first (and in this case, only) element
   - Returns a single integer value

**Why `.values[0]`?**  
Without it, we'd get a Series. With it, we get just the number we want.

````python
chipo[chipo['item_price'] == chipo['item_price'].max()]['quantity'].values[0]
````

**Expected Output:**
```
15
```

**Alternative view: Without `.values[0]`**

**Explanation:**  
Let's see what we get if we stop before extracting the single value.

This returns a pandas **Series** showing:
- The index label (order_id = 1443)
- The column name (quantity)
- The value (15)
- The data type (int64)

**Difference:**  
- With `.values[0]`: Returns `15` (just the number)
- Without it: Returns a Series with metadata

**Which to use?**  
- If you need just the number for calculations → use `.values[0]`
- If you want to see the context (which order_id, etc.) → don't use `.values[0]`

````python
chipo[chipo['item_price'] == chipo['item_price'].max()]['quantity']
````

**Expected Output:**
```
order_id
1443    15
Name: quantity, dtype: int64
```

**See the complete row**

**Explanation:**  
This shows the **entire row** for the most expensive order, giving us full context.

**Benefits of seeing the full row:**
- We can see what item it was (Chips and Fresh Tomato Salsa)
- We see the quantity (15)
- We see the total price ($44.25)
- We see the unit price ($2.95 each)

This is more informative than just getting the quantity number alone.

````python
chipo[chipo['item_price'] == chipo['item_price'].max()]
````

**Expected Output:**
```
          quantity                     item_name choice_description  item_price  \
order_id                                                                          
1443            15  Chips and Fresh Tomato Salsa                NaN       44.25   

          item_price_unit  
order_id

```

### Step 11 (Alternative). Using idxmax with .loc

**Approach 2: Using idxmax() with .loc[]**

**Explanation:**  
This is an alternative approach that uses `.idxmax()` to find the index and then `.loc[]` to access specific data.

**Breaking down the code:**

1. **`chipo['item_price'].idxmax()`**:
   - Finds the index (order_id) with the maximum price
   - Returns: 1443

2. **`chipo.loc[1443, 'quantity']`**:
   - Uses `.loc[]` to access a specific cell
   - Row: 1443 (the index from idxmax)
   - Column: 'quantity'
   - Returns the quantity value at that location

**Important observation:**  
This example shows that `.loc[]` with a single index that appears multiple times (order_id 1443 has multiple items) will return **all rows** with that index, not just one.

**Result:**  
Returns a Series with all quantities from order 1443:
- 15 (Chips and Fresh Tomato Salsa - the expensive one)
- 7 (another item)
- 1, 3, 4, 3, 2 (other items in the same order)

**Key learning:**  
This demonstrates that order_id 1443 was a **large order with multiple items**, and the $44.25 chips were just one part of it!

````python
chipo.loc[chipo['item_price'].idxmax(), 'quantity']
````

**Expected Output:**
```
order_id

1443     7
1443     1
1443     3
1443     4
1443     3
1443     2
Name: quantity, dtype: int64
```

**Analysis:**  
Order 1443 had 7 different items with various quantities. The 15 Chips and Fresh Tomato Salsa was the most expensive line item in this large order.

## Counting Specific Items

### Step 12. How many times was a Veggie Salad Bowl ordered?

This combines **filtering** and **counting** to answer the question.

**Breaking down the code:**

1. **`chipo['item_name'] == 'Veggie Salad Bowl'`**:

   - True where item_name is exactly 'Veggie Salad Bowl'

2. **`chipo[...]`**:

## Multiple Conditions

### Step 13. How many times did someone order more than one Canned Soda?

**Explanation:**  
This uses **multiple conditions** with the AND operator (`&`) to filter data.

**Breaking down the code:**

1. **`chipo['item_name'] == 'Canned Soda'`**:
   - First condition: item must be 'Canned Soda'

2. **`chipo['quantity'] > 1`**:
   - Second condition: quantity must be greater than 1

3. **`&`** operator:
   - Combines conditions with AND logic
   - Both conditions must be True
   - **Important:** Parentheses are required around each condition when using `&`

4. **`len()`**:
   - Counts the number of rows that meet both conditions

**Logical operators in pandas:**
- **`&`**: AND (both conditions must be True)
- **`|`**: OR (at least one condition must be True)
- **`~`**: NOT (inverts True/False)

**Common mistake:**  
Don't use Python's `and`/`or` keywords - use `&`/`|` operators instead for pandas.

**Result interpretation:**  
There were **20 instances** where someone ordered multiple Canned Sodas in a single line item.

````python
len(chipo[(chipo['item_name'] == 'Canned Soda') & (chipo['quantity'] > 1)])
````

**Expected Output:**
```
20
```

**Alternative Method 1: Using .shape[0]**

**Explanation:**  
`.shape` is an attribute that returns the dimensions of the DataFrame as a tuple: `(rows, columns)`

**Breaking down `.shape[0]`:**
- `.shape` returns something like `(20, 5)` meaning 20 rows and 5 columns
- `[0]` gets the first element of the tuple → the number of rows
- This is equivalent to `len()` but more explicit about getting row count

**When to use:**
- `.shape[0]` is clearer when you explicitly want the row count
- `len()` is more concise and commonly used
- Both give the same result

**Note:**  
`.shape[1]` would give you the number of columns (5 in this case).

````python
chipo[(chipo['item_name'] == 'Canned Soda') & (chipo['quantity'] > 1)].shape[0]
````

**Expected Output:**
```
20
```

Same result as using `len()` - confirming 20 orders.

**Alternative Method 2: Using .count()**

**Explanation:**  
`.count()` returns the count of non-null values in each column.

**What you'll see:**
- All columns show the same count (20) because we're counting complete rows
- Returns a Series with counts for each column

**Comparison of the three methods:**

1. **`len(filtered_dataframe)`**:
   - Returns: Single integer (20)
   - Pros: Simple, concise, fast
   - Best for: When you just need the count

2. **`.shape[0]`**:
   - Returns: Single integer (20)
   - Pros: Explicit about getting row count
   - Best for: Code clarity, when working with dimensions

3. **`.count()`**:
   - Returns: Series with count per column
   - Pros: Shows if any columns have missing data
   - Best for: Data quality checks, seeing column-by-column counts

**Best practice:**  
For simple "how many rows" questions, use `len()` or `.shape[0]`.

````python
chipo[(chipo['item_name'] == 'Canned Soda') & (chipo['quantity'] > 1)].count()
````

**Expected Output:**
```
quantity              20
item_name             20
choice_description    20
item_price            20
item_price_unit       20
dtype: int64
```

All columns show 20, confirming 20 orders met both conditions.

## Summary: Key Concepts Learned

### Data Loading and Preparation
- Reading TSV files with `pd.read_csv(sep='\t')`
- Setting index columns with `index_col`

- Creating calculated columns

### Exploring Data
- `.head()` and `.tail()` for quick inspection

- `len(df)` - total rows
- `.count()` - non-null values per column
- `.shape[0]` - number of rows explicitly

### Sorting
- `.sort_values(by='column')` for organizing data

### Best Practices
1. **Always inspect your data** with `.head()`, `.info()` before analysis
2. **Convert data types** when necessary (like removing '$' from prices)
3. **Use descriptive variable names** (chipo, not df)
4. **Choose the right counting method** based on your needs
5. **Use parentheses** when combining conditions with `&` or `|`
6. **Verify your results** using multiple methods when learning

### Common Patterns
```python
# Filter and count
df[df['column'] == value].count()

# Multiple conditions
df[(condition1) & (condition2)]

# Find max/min row
df.loc[df['column'].idxmax()]

# Extract single value
```
