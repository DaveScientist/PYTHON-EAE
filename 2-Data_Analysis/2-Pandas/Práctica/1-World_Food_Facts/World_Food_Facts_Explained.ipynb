## Data Access Methods in Pandas

Now we'll explore **three different ways** to access data in a DataFrame:  
1. **`.loc[]`** - Label-based indexing
2. **`.iloc[]`** - Integer position-based indexing  
3. **Bracket notation `[]`** - Direct column/row access

We'll access the same data using all three methods to understand their differences.

### Method 1: Using .loc[] (Label-based indexing)

**Explanation:**  
`.loc[]` is used for **label-based** indexing - you specify the actual index labels and column names.  

**Syntax:** `dataframe.loc[row_label, column_label]`  

**In this example:**
- `18` is the **index label** (row number)
- `'product_name'` is the **column name**
- This returns the product name at row 18

**Key characteristics:**
- Uses actual labels from the index and column names
- **Inclusive** on both ends when slicing (e.g., `food.loc[10:20]` includes both 10 and 20)
- More readable and explicit
- **Best practice:** Use `.loc[]` when you know the label names

```python
food.loc[18, 'product_name']
```

### Method 2: Using .iloc[] (Integer position-based indexing)

**Step 2a: Get the integer position of the column**

**Explanation:**  
`.columns.get_loc()` returns the **integer position** of a column by its name.  

**In this example:**
- `'product_name'` is the column we're looking for
- This returns the numeric position of that column (e.g., 7 means it's the 8th column, counting from 0)
- We store this position in the variable `pos_product_name` for use in the next step

**Why do we need this?**  
`.iloc[]` only works with integer positions, not column names. So we first need to find the position.

```python
pos_product_name = food.columns.get_loc('product_name')
```

**Step 2b: Access data using integer positions**

**Explanation:**  
`.iloc[]` is used for **integer position-based** indexing - you specify numeric positions only.  

**Syntax:** `dataframe.iloc[row_position, column_position]`  

**In this example:**
- `18` is the **row position** (19th row, counting from 0)
- `pos_product_name` is the **column position** we calculated above
- This returns the same value as `.loc[18, 'product_name']`

**Key characteristics:**
- Uses only integer positions (0-based indexing)
- **Exclusive** on the end when slicing (e.g., `food.iloc[10:20]` includes 10-19, but NOT 20)
- More flexible for programmatic access
- **Best practice:** Use `.iloc[]` when you need to access data by position, especially in loops

**Key difference from .loc[]:**  
- `.loc[]` uses **labels** → `food.loc[18, 'product_name']`
- `.iloc[]` uses **positions** → `food.iloc[18, 7]`

```python
food.iloc[18, pos_product_name]
```

### Method 3: Using bracket notation (Direct access)

**Explanation:**  
Bracket notation `dataframe['column'][row]` provides direct access to columns and rows.  

**Syntax breakdown:**
- `food['product_name']` first selects the entire 'product_name' column (returns a pandas Series)
- `[19]` then selects the value at index position 19 from that Series

**Key characteristics:**
- Two-step process: first get the column, then get the row
- Simpler syntax for single column access
- Uses the index label for the row (19 in this case)

**Important notes:**
- This is **chained indexing** which pandas generally discourages for assignments
- For reading data (like here), it's fine
- For setting values, always use `.loc[]` or `.iloc[]` to avoid warnings

**When to use:**
- Quick column access: `food['product_name']`
- Reading single values: `food['product_name'][19]`
- **Avoid for:** Setting values (use `.loc[]` instead)

```python
food['product_name'][19]
```

---
## Summary: Comparison of Access Methods

| Method | Syntax | Uses | Best For | Slicing Behavior |
|--------|--------|------|----------|------------------|
| **`.loc[]`** | `df.loc[row_label, col_label]` | Labels (names) | When you know exact labels | **Inclusive** both ends |
| **`.iloc[]`** | `df.iloc[row_pos, col_pos]` | Integer positions | Programmatic access, loops | **Exclusive** end |
| **`[]`** | `df['col'][row]` | Mixed (column name, then index) | Quick single column access | Depends on context |

### Key Takeaways:

1. **`.loc[]` is explicit and clear** - use it when you know the names of your rows/columns
2. **`.iloc[]` is position-based** - use it when you need to access data by numeric position
3. **Bracket notation is simple** - good for quick column access, but avoid for setting values
4. **For assignments, always use `.loc[]` or `.iloc[]`** to avoid pandas warnings

### Remember:
- Python uses **0-based indexing** (first element is at position 0)
- `.loc[]` slices are **inclusive**: `[0:5]` includes 0, 1, 2, 3, 4, **and 5**
- `.iloc[]` slices are **exclusive**: `[0:5]` includes 0, 1, 2, 3, 4, **but not 5**
