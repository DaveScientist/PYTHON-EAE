### Step 9. Display a subset of columns (excluding last 3)

**Explanation:**  
`.iloc[]` is used for **integer position-based** column selection.

**Breaking down the code:**  
`euro12.iloc[:, 0:-3]`

1. **`:`** (before comma) - **Row selector**:
   - Select **all rows**
   - Empty slice means "from start to end"

2. **`,`** - Separator between rows and columns

3. **`0:-3`** (after comma) - **Column selector**:
   - `0` = start from the **first column** (index 0)
   - `-3` = go up to (but **not including**) the **3rd column from the end**
   - This **excludes the last 3 columns**

**What gets excluded:**
- Column 32: 'Subs on'
- Column 33: 'Subs off'
- Column 34: 'Players Used'

**Result:**  
Shows all 16 teams with columns 0 through 31 (32 columns total, since 35 - 3 = 32)

**Understanding negative indexing:**
- `-1` = last column
- `-2` = second-to-last column
- `-3` = third-to-last column
- `0:-3` = "from first column up to (not including) third-to-last"

**Alternative ways to accomplish the same:**
- `euro12.iloc[:, :32]` - Explicitly select first 32 columns
- `euro12.drop(columns=['Subs on', 'Subs off', 'Players Used'])` - Drop by name
- `euro12[euro12.columns[:-3]]` - Using column names with slicing

**When to use `.iloc[]` for columns:**
- When you know the position but not the name
- When selecting ranges of columns
- When working with dynamic column positions

````python
euro12.iloc[:, 0:-3]
````

## Advanced Filtering: Working with Multiple Teams

### Step 10. Present only the Shooting Accuracy from England, Italy and Russia

**Explanation:**  
This combines multiple pandas techniques to filter specific rows and columns.

**Breaking down the code:**  
`euro12.loc[euro12.Team.isin(['England', 'Italy', 'Russia']), ['Shooting Accuracy']]`

**Part 1: Row filtering with `.isin()`**  
`euro12.Team.isin(['England', 'Italy', 'Russia'])`

- **`.isin()`** checks if each value in 'Team' column is in the provided list
- Returns a **boolean mask** (True/False for each row)
- True for rows where Team is 'England', 'Italy', or 'Russia'
- False for all other teams

**Why use `.isin()` instead of multiple conditions?**

**Without `.isin()` (verbose):**
```python
(euro12['Team'] == 'England') | (euro12['Team'] == 'Italy') | (euro12['Team'] == 'Russia')
```

**With `.isin()` (concise):**
```python
euro12['Team'].isin(['England', 'Italy', 'Russia'])
```

**Benefits of `.isin()`:**
- Much shorter and more readable
- Works with any number of values
- More efficient for many values
- Less error-prone (no need to remember `|` operators)

**Part 2: Using `.loc[]` for combined selection**  
`euro12.loc[row_condition, column_list]`

- **Before comma**: Row selector (the boolean mask from `.isin()`)
- **After comma**: Column selector (`['Shooting Accuracy']`)
- Returns only the intersection: selected rows × selected columns

**Part 3: Column selection**  
`['Shooting Accuracy']`

- Must be a **list**, even for a single column
- **With brackets** `['Shooting Accuracy']` → Returns a DataFrame
- **Without brackets** `'Shooting Accuracy'` → Would return a Series

**Result:**  
A DataFrame with:
- **3 rows** (England, Italy, Russia)
- **1 column** (Shooting Accuracy)

**Alternative syntax using bracket notation:**
```python
euro12[euro12['Team'].isin(['England', 'Italy', 'Russia'])][['Shooting Accuracy']]
```
This works but `.loc[]` is preferred for clarity and to avoid chained indexing warnings.

````python
acierto_disparo = euro12.loc[euro12.Team.isin(['England', 'Italy', 'Russia']), ['Shooting Accuracy']]
acierto_disparo
````


**Interpretation:**
- Row 3 (England): 50.0% shooting accuracy
- Row 7 (Italy): 43.0% shooting accuracy  
- Row 12 (Russia): 22.5% shooting accuracy

**Note:** The index numbers (3, 7, 12) are preserved from the original DataFrame, showing which rows these teams were in.

## Summary: Key Concepts Learned

### Data Loading
- Loading CSV files directly from URLs
- No need to download files locally first
- Pandas handles HTTP requests automatically

### Display Options
- `pd.set_option('display.max_columns', None)` for showing all columns
- Useful for wide datasets
- Other options: max_rows, width, precision

### DataFrame Inspection
- `.head(n)` - First n rows
- `.tail(n)` - Last n rows
- `.info()` - Detailed structure information
- `.shape` - Dimensions (rows, columns)
- `.columns` - All column names

### Advanced Selection Techniques

**1. Position-based column slicing with `.iloc[]`:**
```python
df.iloc[:, 0:-3]  # All rows, columns except last 3
```
- Use negative indexing to count from the end
- Colon (`:`) means "all" for that dimension

**2. Multiple value filtering with `.isin()`:**
```python
df[df['column'].isin([value1, value2, value3])]
```
- Much cleaner than multiple OR conditions
- Works with any number of values
- More efficient and readable

**3. Combined row and column selection with `.loc[]`:**
```python
df.loc[row_condition, ['column1', 'column2']]
```
- Filter rows and columns in one operation
- First argument: row selector (boolean mask)
- Second argument: column selector (list of names)

### Comparison: `.loc[]` vs `.iloc[]`

| Feature | `.loc[]` | `.iloc[]` |
|---------|----------|----------|
| **Selection method** | Label-based | Position-based |
| **Row selection** | Index labels | Integer positions |
| **Column selection** | Column names | Integer positions |
| **Slicing** | Inclusive | Exclusive end |
| **Use case** | When you know names | When you know positions |

### Best Practices

1. **Use `.isin()` for multiple value filtering**
   - Cleaner than chaining OR conditions
   - More maintainable as lists grow

2. **Use `.loc[]` for combined selection**
   - Explicit about what you're selecting
   - Avoids chained indexing warnings
   - Better performance

3. **Set display options appropriately**
   - Show all columns for wide datasets during exploration
   - Reset or limit for cleaner output in production

4. **Always inspect your data first**
   - Use `.head()`, `.info()`, `.columns`
   - Understand structure before filtering
   - Check for missing values and data types

### Common Patterns

```python
# Filter multiple values
df[df['column'].isin([val1, val2, val3])]

# Select specific rows and columns
df.loc[condition, ['col1', 'col2']]

# Exclude last n columns
df.iloc[:, :-n]

# Select range of columns by position
df.iloc[:, start:end]
```

### Key Takeaways

- **`.isin()`** is your friend for filtering multiple values
- **`.loc[]`** is best for label-based selection
- **`.iloc[]`** is best for position-based selection
- **Negative indexing** (`-3`, `-1`) counts from the end
- **Display options** can make data exploration much easier
- **Always verify** your selections by checking shape and head
