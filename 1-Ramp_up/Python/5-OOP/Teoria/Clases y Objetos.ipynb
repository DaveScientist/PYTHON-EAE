## 1. Classes
Classes are the way we have to describe objects. So far we have seen basic classes that are included in Python like *int*, *str* or somewhat more complex classes like *dict*. But, **what if we want to create our own objects?** In object-oriented languages we have the possibility to define new objects that are more similar to our use cases and make programming easier to develop and understand.

**An integer is an object of the *int* class that has different characteristics than text**, which is of the *str* class. For example, **how do we know that a car is a car?** what characteristics does it have? Cars have a brand, a number of horsepower, some are automatic, others are not… In this way we translate to machine language, to programming, a concept that we have very clear and internalized.
 
So far, we have seen several classes, for example the *str* class. When we saw the data type, Python printed `str` on the screen. And being `str`, it had properties that other objects didn't have, like the functions .upper() or .lower().

The syntax to create a class is:
```Python
class ClassName:
    # Things of the class
```

Normally for the class name we use *CamelCase*, which means that it is defined in lowercase, without spaces or hyphens, and playing with uppercase letters to differentiate words.

Look at the [*built_in* *String* class](https://docs.python.org/3/library/stdtypes.html#str)

````python
class Car:
    # things of the class
    pass
````

The `pass` statement is used to force the end of the *Car* class. We have declared it, but it carries nothing. Python demands a definition of the class and we can ignore that demand through the `pass` statement.

````python
print(type(Car))
print(type(str))
````

Well, Car is of type `type`, of course because **it's not an object as such**, but it is a class. When we create cars, these will be of class *Car*, that is, of type *Car*, so it makes sense that *Car* is of type `type`.

### Class vs Object
**The class is used to define something**. Just like with functions. We create the skeleton of what will be an object of that class. Therefore, **once we have the class defined, we will instantiate an object of that class**. It's like creating the concept of a car, with all its characteristics and functionalities. Later, throughout the program, we can create car-type objects, which conform to what is defined in the car class. Each car will have a brand, a power, etc...

````python
first_car = Car()
print(first_car)
print(type(first_car))
````

````python
mi_lista = list()
mi_booleano = bool()
mi_int = int()


print(mi_lista)
print(mi_booleano)
print(mi_int)
````

Now we have an object of type Car, which is called `primer_coche`. When we print its type, we see that it is of type Car, and when you print the object to the screen, it simply tells us its type and an identifier.

We can create as many cars as we want

````python
citroen = Car()
seat = Car()

print(citroen)
print(seat)

# they are two different objects

citroen == seat

````

So far all our cars are the same, we haven't defined the class well, so it will be difficult to differentiate one car from another. Let's see how to achieve that differentiation.

## 2. Attributes
They are **the characteristics that define the objects of a class**. The brand, the color, power of the car. These are attributes, which are defined generically in the class and then each *Car* object will have a value for each of its attributes.

We define the attributes after the class declaration. And then they are accessed using the syntax `object.attribute`

Let's start defining attributes in the cars.

````python
class Car:
    doors = 4
    wheels = 4
    engine = "gasoline"
````

Now all the cars we create will have 4 doors and 4 wheels.

````python
citroen = Car()

print(citroen)
print(citroen.doors)
print(citroen.wheels)

seat = Car()

print(seat)
print(seat.doors)
print(seat.wheels)
````

We can also modify the attributes. Python makes this very easy, we change them directly by reassigning values. In other programming languages this has to be implemented through methods called `getters` and `setters`.

````python
citroen = Car()
citroen.doors = 2
print(citroen)
print(citroen.doors)
````

## 3. Constructor
When we create an object of the *Car* class, we have to define it well to differentiate it from other cars. That initial definition is done in the constructor of the class. They are input arguments that the object asks us to define that instance differently from other instances of the same class.

**How do we define this?** Through the `__init__` statement, inside the class.

````python
class Car:
    doors = 4
    wheels = 4
    
    def __init__(self, brand_car):
        self.brand = brand_car
````

In the constructor declaration we have put the word `self`. **We will have to put it always**. It refers to the own instance of car, that is, when we create new cars.

In this case we are differentiating the common attributes of the *Car* class, from the particular attributes of the cars, such as the brand. That is why the brand goes together with `self`, because it doesn't refer to the generic car class, but to each car we create.

````python
citroen = Car("Citroen")

# print(citroen)
# print(citroen.doors)
# print(citroen.wheels)
print(citroen.brand)
````

Now we can differentiate cars by their brand. To access the brand attribute, we do it the same way as with the others.

````python
citroen = Car("Citroen")
seat = Car("Seat")
renault = Car ("Renault")

print(citroen.brand)
print(seat.brand)
print(renault.brand)
````

Now we can solve the issue that not all cars have 4 doors

````python
class Car:
    wheels = 4
    
    def __init__(self, brand_car, number_doors):
        self.brand_car = brand_car
        self.number_doors = number_doors
````

## 4. Methods
They are functions that we can define within classes. These functions will change the state of some attribute or perform calculations that serve as output. A simple example can be, a method of the Car class that outputs the power in kilowatts, instead of in horsepower. Or if it has a maintenance state (ITV passed or not), it modifies that state.

The constructor is a type of method. The difference with the rest of the methods lies in its name, `__init__`. The syntax for defining methods is as if it were a function. And then to call the method we use `object.method(method_arguments)`. We have already used this before, when we did a `string.lower()`, we simply called the `lower()` method, which required no arguments, from the *string* class.

````python
class Car:
    wheels = 4
    
    def __init__(self, brand_car, number_doors):
        self.brand_car = brand_car
        self.number_doors = number_doors
        
    def characteristics(self):
        return "Brand: " + self.brand_car + ". Number of Doors: " + str(self.number_doors) + ". Number of Wheels: " + str(self.wheels)
````

````python
my_car = Car("Audi", 2)
my_characteristics = my_car.characteristics()
print(my_characteristics)
````

Notice that to call the wheels we use `self`, even though we hadn't put it in the constructor. This way we avoid calling another variable in the program that is called *wheels*. We make sure that they are the wheels of that car with the `self`.

## 5. Documentation
Just like with functions, in classes we can also document with the *built-in* `__doc__` method. It is a method of `class`. Therefore, we can put at the beginning of the class a documentation with everything this class does. The same happens with the methods of the class. It is recommended to give a brief definition of the functionalities of classes/methods and describe what the inputs and outputs of the methods are like. What it expects to receive and what type.

````python
class Car:
    '''
    Car class used as an example for the class
    Parameters:
        brand_car: distinguishes, manufacturer of the car
        number_doors: there are cars with 2 and 4 doors
    '''
    wheels = 4
    
    def __init__(self, brand_car, number_doors):
        '''
        Documentation of the init
        '''
        self.brand_car = brand_car
        self.number_doors = number_doors

print(Car.__doc__)
print(Car.__init__.__doc__)
````
