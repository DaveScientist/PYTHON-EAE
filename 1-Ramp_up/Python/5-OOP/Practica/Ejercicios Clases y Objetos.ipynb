## Exercise 1
We are a company that has several appliance stores. We need a program to manage the stores, see the sales they have had, how many employees there are, etc... For this we are going to model the store in a class, which will have the following characteristics:
* Class name: "Store"
* Common attributes:
    * Type: "Appliances"
    * Open: True
* Attributes specific to each store:
    * Name - String
    * Address - String
    * Number of employees - int
    * Sales last 3 months - List of 3 numbers
* Method to calculate sales from all months, returning a number with all sales.
* Method that calculates the average sales of the last months, per employee
* Method that returns in a string the name of the store, along with its address.
* Method that gets the sales from the last month.
* Method to give a sales projection in case X money has been invested in marketing for that store. Where X is the input parameter. If X is less than 1000, sales for the last 3 months would have been (1.2 x sales), if greater than or equal to 1000, sales would have been (1.5 x sales). The method rewrites the sales attribute for the last 3 months, and also returns the sales with the increase.

It is requested:
1. Implement the "Store" class
2. Create three stores with invented data
3. Check in at least one of them, everything implemented in the store class (its attributes, average sales, sales/employee...)
3. Calculate the sales from the last month of all stores. For this use the `for` loop
4. Print on screen the names of the stores whose address contains the string "Avenue"
5. I'm sure you've already done this section... Document the class :)

```python
class store:
    """
    Represents a retail store with sales and employee data.

    Attributes:
        type (str): Class attribute representing the store category.
        open (bool): Class attribute indicating if stores are open.
        name (str): Store name.
        address (str): Store physical address.
        employees (int): Number of employees.
        sales (list): List containing sales for the last 3 months.
    """

    # Class attributes (shared across all instances)
    type = "Appliances"
    open = True

    def __init__(self, name, address, number_of_employees, sales_last_3_months):
        """
        Initializes a new Store instance.

        Args:
            name (str): Name of the store.
            address (str): Store address.
            number_of_employees (int): Total number of employees.
            sales_last_3_months (list): Sales values for the last three months.
        """
        self.name = name
        self.address = address
        self.employees = number_of_employees
        self.sales = sales_last_3_months

    def sum_of_sales(self):
        """
        Calculates the total sales of the last three months.

        Returns:
            int/float: Sum of sales.
        """
        return sum(self.sales)

    def sales_per_employee(self):
        """
        Calculates the average sales generated per employee.

        Returns:
            float: Sales per employee.
        """
        return sum(self.sales) / self.employees

    def characteristics(self):
        """
        Returns a formatted string with store basic information.

        Returns:
            str: Store name and address.
        """
        return "Name: " + str(self.name) + ", Address: " + str(self.address)

    def last_month_of_sales(self):
        """
        Returns the sales value of the most recent month.

        Returns:
            int/float: Last month's sales.
        """
        return self.sales[2]

    def marketing_projection(self, x):
        """
        Estimates projected sales based on a marketing budget threshold.

        Args:
            x (int/float): Marketing investment amount.

        Returns:
            float: Projected total sales.
        """
        if x < 1000:
            return sum(self.sales) * 1.2
        else:
            return sum(self.sales) * 1.5
```

```python
# --- Creating store instances ---
store1 = store("Zara", "Calle Araucaria", 100,[55000,15000,25000])
store2 = store("Mango", "Calle Principe de Vergara", 70,[45000,50000,15000])
store3 = store("Gucci", "Avenue Bravo Murillo", 50,[35000,10000,15000])
```

```python
#Outputs containing every single attribute of one of the stores:
print(store1.type)
print(store1.open)
print(store1.name)
print(store1.address)
print(store1.employees)
print(store1.sales)
print(store1.sum_of_sales())
print(store1.sales_per_employee())
print(store1.characteristics())
print(store1.last_month_of_sales())
print(store1.marketing_proyection(900))
```

```python
# List containing all store objects
stores = [store1, store2, store3]


# --- Calculating total sales of the last month across all stores ---

total_sales = 0
for store in stores:
    total_sales += store.last_month_of_sales()

print(total_sales)
```

```python
def find_store_by_address(stores, keyword):
    """
    Searches for a store containing a specific keyword in its address.

    Args:
        stores (list): List of Store objects.
        keyword (str): Word to search for in the address.

    Returns:
        str: Store name if found, otherwise a not-found message.
    """
    for store in stores:
        if keyword.lower() in store.address.lower():
            return store.name
    return "There are no stores under that address"

# Example usage
print(find_store_by_address(stores, "avenue"))
```

## Exercise 2
In this exercise we are going to implement a *Dog* class in Python. The class has the following characteristics:
* Things we know for sure a dog has
    * It has 4 legs
    * 2 ears
    * 2 eyes
    * A speed of 0. By default, the dog is stopped
* When initialized:
    * The dog will be of a certain breed
    * By default it will have "Brown" fur, unless otherwise stated.
    * By default it will have no owner, unless otherwise stated.
    
* It will also have a method called walk, which has an input argument (speed_increase). This value will be added to the speed the dog already had.
* It needs another method (stop), where we will set the speed to 0.
* Another method called "bark", which will have an input argument, and the output will be the following string: "WOOF!" + the input argument.


It is requested:
* Implement the *Dog* class
* Create an object of type *Dog*, without an owner
* Check that all its attributes are correct
* Test that it walks, and check its speed
* Stop it
* Document the *Dog* class

```python
class Dog:
    """
    Represents a Dog with physical characteristics and behaviors.

    Attributes:
        legs (int): Number of legs. Always 4.
        ears (int): Number of ears. Always 2.
        eyes (int): Number of eyes. Always 2.
        speed (int/float): Current speed of the dog. Default is 0.
        breed (str): Breed of the dog.
        fur_color (str): Color of the fur. Default is "Brown".
        owner (str or None): Owner of the dog. Default is None.
    """

    # Constant physical attributes (shared by all dogs)
    legs = 4
    ears = 2
    eyes = 2

    def __init__(self, breed, fur_color="Brown", owner=None):

        """
        Initializes a Dog instance.

        Args:
            breed (str): Breed of the dog.
            fur_color (str, optional): Color of the fur. Defaults to "Brown".
            owner (str, optional): Owner of the dog. Defaults to None.
        """
        self.breed = breed
        self.fur_color = fur_color
        self.owner = owner
        self.speed = 0  # Dog starts stopped

    def walk(self, speed_increase):
        """
        Increases the dog's speed.

        Args:
            speed_increase (int/float): Value to add to current speed.
        """
        self.speed += speed_increase

    def stop(self):
        """
        Stops the dog by setting its speed to 0.
        """
        self.speed = 0

    def bark(self, sound):
        """
        Returns a barking sound combined with an input string.

        Args:
            sound (str): Extra string to append after "WOOF!"

        Returns:
            str: Barking string.
        """
        return "WOOF! " + str(sound)


# Creating a Dog object without an owner


dog1 = Dog(breed="Golden Retriever")

# Checking attributes
print("Breed:", dog1.breed)
print("Fur color:", dog1.fur_color)
print("Owner:", dog1.owner)
print("Legs:", dog1.legs)
print("Ears:", dog1.ears)
print("Eyes:", dog1.eyes)
print("Initial speed:", dog1.speed)

# Testing walk

dog1.walk(10)
print("Speed after walking:", dog1.speed)

# Stopping the dog

dog1.stop()
print("Speed after stopping:", dog1.speed)

# Testing bark

print(dog1.bark("I am tired!"))
```
