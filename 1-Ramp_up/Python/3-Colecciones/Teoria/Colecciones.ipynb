## 1. Lists
We already know quite a bit about lists. Let's review what we can do with them, as well as some new features

````python
# Lists of numbers, strings, booleans, with repeated elements, nested lists...
nums = [6, 2, 8, 3, 4, 5, 5]
months = ["January", "February", "March"]
mix = [5, 7, "April", True, None, ["White", "Black"]]

# Length of the list
print(len(nums))

# Indexing
print(months[0])
print(months[len(months)-1])
````

### Access
Between square brackets we introduce the index of the element we want to access

````python
print(months[0])
print(months[-1])
````

If we pose the problem the other way. We have the values of the list and what we want is to get the index of those values within the list

````python
months.index("March")
````

**Slicing**: we use slicing to access several consecutive elements of the list

````python
degrees = [22, 34, 15, 26, 18, 22]

print(degrees[2:4])
print(degrees[2:5])
print(type(degrees))
````

### Modify elements
Lists are mutable, so we can modify them

````python
print(months)

months[2] = "April"
print(months)
````

### Add elements

````python
bikes = ['honda', 'yamaha', 'suzuki']
print(bikes)

bikes.append('ducati')
print(bikes)

bikes.insert(2, 'kawasaki')
print(bikes)
````

````python
cars = ['VW', 'Seat', 'BMW', 'VW']
print(cars)

print(cars.remove('VW'))
print(cars)
````

## 2. Tuples
Very similar to lists. If in previous Notebooks we defined lists as ordered and mutable, tuples are ordered and immutable.

* **Immutable**: once the tuple is created with its initial elements, it cannot be modified
* **Ordered**: we can access its elements through the index and reorder the tuple as we wish

Tuples are usually used for small collections of data that won't change throughout the program, such as constants.

If lists were constructed using square brackets `[]`, tuples are constructed with parentheses `()`

````python
first_tuple = (3, "text")
print(first_tuple)
print(type(first_tuple))
````

Parentheses are already used to reorder operations `5 * (3 + 4)`, so you have to add a comma when defining a tuple with a single element, otherwise Python will interpret it as a number.

````python
another_tuple = (1,)
print(another_tuple)
print(type(another_tuple))
````

````python
nested_tuple = (1, 2, 3, ("A", "B", "C"))
print(nested_tuple)

print(nested_tuple[0])

for i in nested_tuple:
   print(i)

print(nested_tuple[0:2])

print(len(nested_tuple))
````

## 3. Dictionaries
In this type of collections we can not only group data, but we also have the possibility to establish a relationship between them, with the form **key: value**.

The syntax is using curly braces `{}`

> dictionary_name = {key1: value1, key2: value2, key3: value3}

If we saw that lists are mutable and ordered, and tuples immutable and ordered, dictionaries are:

* **Mutable**: we can modify its elements
* **Unordered**: the power of accessing dictionaries is their key-value format, not their order. Even so, the dictionary preserves the order of the elements as we insert them.

Let's declare a dictionary in which we will define elements of a car

````python
car = {"colour": "Gray","brand": "VW", "weight": 1500, "Automatic transmission": True}
car
````

````python
print(type(car))
````

````python
car['weight']
````

You can store all types of data: number, string, boolean, lists, tuples, dictionaries...

When we declare the dictionary, we can do it in a single line `car = {"color"="Gray", "brand"...}`, although it is recommended for simplicity when reading the code, to perform this action on several lines. If we write the first element and press enter, Jupyter automatically introduces the indentation.

### Access, modify, add, delete

````python
# Access
print(car['Automatic transmission'])

# # Modify
car['colour'] = "White"
print(car)

# # Add
car['doors'] = 4
print(car)

# # Delete
del car['doors']
print(car)
````

````python
print(car.keys())
print(car.values())
````

## 4. Sets
Another *built-in* collection in Python, composed of a set of *keys*. Very similar to dictionaries. They have the following characteristics:
* **Mutable**: we can modify them once they have been created
* **Unordered**
* **Unique elements**: it consists of a set of unique keys

**When to use sets?** When we're looking for uniqueness in our data and we don't care about the order.

I leave you here [the documentation](https://docs.python.org/2/library/sets.html) to make queries about sets

Its syntax is:
>```Python
> my_set = {element1, element2, element3}
>```

It has a very similar syntax to dictionaries, but in this case the `:` are not used.

````python
colours = {"red", "blue", "white"}
print(type(colours))

colours2 = {"red", "blue", "white", "red"}
print(colours2)

len(colours2)
````

The elements of the set are unique, so if during the declaration, or later adding elements there was any duplicate, the set would ignore it. In fact, when you apply `len`, it shows the number of unique values it has.

````python
days = {'day', 4}
print(days)

days = set()
days.add("Monday")
days.add("Tuesday")
print(days)

days.remove("Tuesday")
print(days)

print("Monday" in days)
print("Friday" not in days)

for i in days:
    print(i)
````

## 6. List comprehension

Lists can also be created by comprehension (in English, *comprehension*). This has much to do with the definition of sets by comprehension from mathematics. For example, in mathematics this expression:
$$ S = \{x^2 |\ x \in [3, 7), x \  odd\}$$
This expression defines a set $ S = \{9, 25\}$, since the interval is open on the right, so it only comprises the numbers 3, 4, 5 and 6, but the condition says that we only take into account the odd numbers, that is, 3 and 5, and the expression at the beginning says that we take the squares. Remembering that to know if a number is even or odd we use the modulo operator in Python (`%`), this can be written (defining a list `s` instead of a set) as:

````python
list_comp = [x**2 for x in range(3, 7) if x % 2 == 1]
print(list_comp)
````

````python
list_comp = []
for x in range(3, 7):
    if x % 2 == 1:
        list_comp.append(x**2)
````

If we have a piece of code that defines a list like this:
``` python
new_list = []
for e in old_list:
    if filter(e):
        new_list.append(transformation(e))
```
It can be written, using comprehension, as:
```python
new_list = [transformation(e) for e in old_list if filter(e)]
```

The definition of lists by comprehension is very convenient, compact and expressive. However, we should not overuse it; if the expression we write is very complex it can become very difficult to read. In that case it would be preferable the "classic" definition of the list.

Let's see this with an example. Suppose we have a list of `ages`, from which we want to obtain a new list, `adults` with those that are 21 or older. We would write:

````python
ages = [19, 20, 18, 21, 19, 25]
print(ages)
adults = [x for x in ages if x >= 21]
print(adults)
````

## 7. Summary

````python
# Lists
print("Lists")
nums = [6, 2, 8, 3, 4, 5, 5]

# Size
print(len(nums))

# Access
print(nums[2])
print(nums[-1])

# Access by index
print(nums.index(5))
nums.append('numbers')
print(nums)

# Delete
nums.remove("numbers")
print(nums)

# Sort
nums.sort()
print(nums)

# Size of the list
print(len(nums))

# Reverse order
nums.reverse()
print(nums)

# Count of a value
print(nums.count(22))
````

````python
# Tuples
first_tuple = (3, "text")

# Nesting of tuples
nested_tuple = (1, 2, 3, ("A", "B", "C"))
print(nested_tuple)

# Access is like in lists
print(nested_tuple[0])

# They are also iterable elements
for i in nested_tuple:
    print(i)
    
# Slicing also works the same
print(nested_tuple[0:2])

# Length of the tuple
print(len(nested_tuple))
````

````python
# Dictionary
car = {"color": "Gray",
        "brand": "VW",
        "weight": 1500,
        "Automatic transmission": True}

print(car)

car = {"color": "Gray",
        "brand": "VW"
        }

for i in car: # This is going through the keys
    print(i) # We print the key
    print(car[i]) # We print the value
    
# Another way to iterate through the elements
for key, value in car.items():
    print(key, value)
````

````python
# Sets
colors = {"red", "blue", "white"}
print(colors)

# Look what happens when we put duplicates
colors2 = {"red", "blue", "white", "red"}
print(colors2)

# Len also works here
len(colors2)

# Create an empty set. It's not created with {}, because if not, it would be a dictionary.
days = set()
print(days)

# Add elements
days.add("Monday")
days.add("Tuesday")
days.add("Wednesday")
print(days)

# Delete elements
days.remove("Wednesday")
print(days)

# Search for elements
print("Monday" in days)
print("Friday" not in days)

# It's also an iterable
for i in days:
    print(i)
````
