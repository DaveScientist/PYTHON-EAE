## Exercise 12
Convert the following list into a dictionary
> ```ejer_12 = [("x", 1), ("x", 2), ("x", 3), ("y", 1), ("y", 2), ("z", 1)]```

TIP: You can use two control variables in a `for` loop. One for each value of the tuples.

````python
ejer_12 = [("x", 1), ("x", 2), ("x", 3), ("y", 1), ("y", 2), ("z", 1)]
print(ejer_12)
print(dict(ejer_12))

#IA result, EXPLICAR??
resultado = {}

# Usamos dos variables de control: 'clave' para la letra y 'valor' para el número
for clave, valor in ejer_12:
    # Si la clave no está en el diccionario, la inicializamos con una lista vacía
    if clave not in resultado:
        resultado[clave] = []
    
    # Agregamos el valor a la lista correspondiente de esa clave
    resultado[clave].append(valor)

print(resultado)
````
