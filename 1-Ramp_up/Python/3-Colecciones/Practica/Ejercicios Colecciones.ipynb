## Exercise 3
Create a new list with all combinations of the following two lists:
> ```ejer_3_1 = ["Hello", "friend"]```
>
> ```ejer_3_2 = ["How", "are"]```

Get the following output:

['Hello How', 'Hello are', 'friend How', 'friend are']

````python
ejer_3_1 = ["Hello", "friend"]
ejer_3_2 = ["How", "are"]

new_list = []

for word1 in ejer_3_1:
    for word2 in ejer_3_2:
        new_list.append(word1 + " " + word2)

print(new_list)
````

## Exercise 5
Given the following list, delete all values equal to 3
> ```ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]```

TIP: Do not try to delete elements from the list you are iterating. Make a copy with `ejer_5.copy()`.

````python
ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]

for x in ejer_5.copy():
    if x == 3:
        ejer_5.remove(x)

print(ejer_5)
````

## Exercise 13
1. Create an ascending sorted list with the dictionary keys
2. Create another list sorted in descending order with the values
3. Add a new key/value
4. Find the key with value 2 in the dictionary
5. Iterate the key and value of the dictionary with a single for loop
> ```ejer_13 = {4:78, 2:98, 8:234, 5:29}```

````python
ejer_13 = {4:78, 2:98, 8:234, 5:29}

keys_asc = sorted(ejer_13.keys())
print(keys_asc)

values_desc = sorted(ejer_13.values(), reverse=True)
print(values_desc)

ejer_13[6] = 2
print(ejer_13)

for key, value in ejer_13.items():
    if value == 2:
        print(key)

for key, value in ejer_13.items():
    print(key, value)


````

## Exercise 14
Merge both dictionaries. To do this, use `update`
> ```ejer_14_1 = {1: 11, 2: 22}```
>
