## Exercise 5
Given the following list, delete all values equal to 3
> ```ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]```

TIP: Do not try to delete elements from the list you are iterating. Make a copy with `ejer_5.copy()`.

````python
ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]

ejer_5_2 = ejer_5.copy()

for i in ejer_5:
    if i == 3:
        ejer_5_2.remove(3)

    print(ejer_5_2)
````

## Exercise 12
Convert the following list into a dictionary
> ```ejer_12 = [("x", 1), ("x", 2), ("x", 3), ("y", 1), ("y", 2), ("z", 1)]```

TIP: You can use two control variables in a `for` loop. One for each value of the tuples.

````python
ejer_12 = [("x", 1), ("x", 2), ("x", 3), ("y", 1), ("y", 2), ("z", 1)]

print(type(ejer_12))

ejer_12_dict = {}

for i, j in ejer_12:
    ejer_12_dict[i] = j

    print(ejer_12_dict)
````

````python
dic = {}
for i, j in ejer_12:
    x = dic.get(i, [])
    x.append(j)
    dic[i] = x

print(dic)
````

## Exercise 16
Multiply all values of the dictionary
> ```ejer_16 = {1: 11, 2: 22, 3: 33, 4: 44, 5: 55}```

````python
ejer_16 = {1: 11, 2: 22, 3: 33, 4: 44, 5: 55}

prod = 1
for i in ejer_16.values():
    prod = prod * i

print(prod)
````
