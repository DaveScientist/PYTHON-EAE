### Step 6. Sort by the name of the item

```python
chipo.sort_values(by=['item_name'], ascending=False)   # True si lo quieres ascendente, False Descendente.
```

```python
#agrupamos por iten_name y luego la media de cada uno (al final es el mismo precio)
precio_item_name = chipo.groupby('item_name',sort=True)['item_price'].mean()
display(precio_item_name.round(2))
```
