## 1. Functional programming
Depending on the problem we want to solve, we will choose one programming paradigm or another. Let's see the difference between an object-oriented language and a functional one:

* **Object-Oriented Programming (OOP)**: this is the case of Python. In this paradigm we encapsulate program elements into objects, which have a type, **a state**, attributes and behaviors. Object-oriented languages include Java, Python, JavaScript, Ruby, among others.

* **Functional Programming**: The program is divided into a set of functions. Therefore, it is a continuous input/output flow: functions receive an *input*, perform operations, and return an *output*. Functional languages can offer good performance but are sometimes harder to develop because recursion is commonly used. Popular functional languages include [Clojure](https://clojure.org) and [Scala](https://www.scala-lang.org).

So, what is Python? Is it object-oriented...? Is it functional? **Python is object-oriented, but it also includes certain built-in features of functional programming**, such as `map` or `filter`, which give us alternative ways to solve problems and can improve performance.

As you can see, Python lets you mix object-oriented features with other paradigms like functional programming.

How can we iterate over a list and compute the sum of its elements?

````python
lista = [1,2,3,4,5]

suma = 0
for i in lista:
    suma += i
suma
````

Aside from Python's built-in `sum()` function, we also have the option of using recursion, where we avoid loops and perform operations via functions only.

````python
def suma_fun(lista):
    if len(lista) == 0:
        return 0
    return lista[0] + suma_fun(lista[1:])
````

````python
lista = [1,2,3]
suma_fun(lista)
````

The main difference lies in **state**. In the `for` loop example the program's state changes due to variables like `suma` and `i`. In the recursive example there are no changing states: functions call other functions, receive an *input* and return an *output*. We'll explore this further in the notebook.

## 2. Lambda function
Recall the syntax to create a function:
```Python
def function_name(arguments):
    function body
    return output
```
A function has a name, parameters, operations inside and an output. Here we present a quicker way to create simple functions using the `lambda` expression. Lambdas let you create single-expression anonymous functions.

Lambda functions **are anonymous**; they have no name and simply evaluate the expression you declare.

Syntax:
```Python
lambda arguments: expression
```

Let's see an example

````python
# Declare function
def suma(x,y):
    return x + y

# Assign to a variable
suma_var = suma

# Check type
print(type(suma_var))

# Use the function
suma_var(5, 7)
````

As you can see, functions are objects that can be assigned to variables. Now let's translate this into a `lambda` function.

````python
lambda x,y: x + y
````

````python
# If we assign it to a variable
suma_lambda = lambda x,y: x + y

suma_lambda(5,7)
````

Lambda functions are usually used as arguments to higher-order functions — functions that accept other functions as parameters — such as `map`, `reduce` or `filter`.

Lambdas can also make code more readable. Here's an example working with coordinates.

````python
estaciones = ((40.4440297, -3.6956047), (40.4585318, -3.684715))

inicio = lambda x: x[0]
fin = lambda x: x[1]

print(inicio(estaciones))
print(fin(estaciones))
````

````python
lat = lambda x: x[0]
lon = lambda x: x[1]

print(lat(inicio(estaciones)))
print(lon(inicio(estaciones)))
````

## 3. Map
`map` is a built-in function that takes two arguments: a function and an iterable (list, tuple, string, etc.). **It applies the function to each element of the iterable**. It maps every value of the iterable through the function and returns a new iterable with the results.

Syntax:
```Python
map(function, iterable)
```

`map` is a way to replace loop functionality. It is useful when you want to apply an operation to every element of a list.

````python
my_list = [1,5,4,6,11]

# my_list + 2 is not valid for lists

new_list = map(lambda x: x + 2, my_list)

print(list(new_list))
````

Note that `map` returns a map object, which is an iterable that can be converted to a list. We used a simple `lambda` here, but you can pass any function. Here's another example with different operations.

````python
my_tuple = (True, False, True, True)

my_tuple = map(lambda x: not x, my_tuple)

print(list(my_tuple))
````

`map` works with any iterable, so you can also apply it to strings.

````python
my_tuple = ("AAA", "BBB", "CCC")

my_tuple = map(lambda x: "W-" + x, my_tuple)

print(tuple(my_tuple))
````

You can even separate the `lambda` function and reuse it elsewhere.

````python
my_list = ("AAA", "BBB", "CCC")

my_fun = lambda x: "W-" + x

new_list = map(my_fun, my_list)

print(tuple(new_list))
````

**You can also pass your own functions.** Imagine the versatility: inside the function you can perform any calculation and it will be applied to each element of the iterable.

````python
my_list = ("AAA", "BBB", "CCC")

def aniade_w(x):
    return "W-" + x

new_list = map(aniade_w, my_list)

print(tuple(new_list))
````

`map` also works with functions that accept multiple arguments: simply pass additional iterables to `map`. If the iterables have different lengths, `map` stops at the shortest one.

````python
circle_areas = [3.56773, 5.57668, 4.00914, 56.24241, 9.01344, 32.00013]

decimales = range(1,7)
print(list(decimales))

# round(decimal_number, number_of_decimals)
result = map(round, circle_areas, decimales)

print(list(result))
````

## 4. Reduce
`reduce` is not a built-in but is included in the `functools` package. Its name hints at what it does: **it reduces/aggregates data**. Syntax:

```Python
reduce(function, iterable[, initial])
```

Like `map`, `reduce` takes a function and an iterable. The function passed to `reduce` receives two arguments (the first two elements of the iterable), and there is an optional third `initializer` argument to start the operation. While `map` transforms each element, **`reduce` aggregates all elements into a single value**.

````python
from functools import reduce

lista = [1,3,5,6,2]

resultado = reduce(lambda x,y: x+y, lista)

print(resultado)
````

````python
# (1+3) = 4 (4 + 5) = 9 (9+6) = 15 (15+2) = 17
# [1,3,5,6,2]
# [4,5,6,2]
# [9,6,2]
# [15,2]
# [17]
````

````python
lista
````

Think of `reduce` like a recursive function: the inner function takes two arguments (the first two elements), then the result is combined with the third element using the same operation, and so on until the iterable is exhausted.

````python
lista = [1,3,5,6,2]

resultado = reduce(lambda x,y: x+y, lista, 100)

print(resultado)
````

`reduce` accepts an optional third argument called `initializer`. By default it's `None`; if provided, the reduction starts with that value (as if it were the first element).

````python
lista_strings = ["Hola ", " me", " llamo", " Ralph"]

resultado = reduce(lambda x,y: x+y, lista_strings)

print(resultado)
````

## 5. Filter
With this built-in function you can **filter elements from an iterable**. Syntax:

```Python
filter(function, iterable)
```

`filter` is similar to `map` but the function must return a boolean (`True`/`False`) — a mask — and only elements with `True` are kept. Unlike `map`, `filter` only uses a single iterable. You can pass lambdas or your own functions, but ensure the output is `True` or `False`.

Example:

````python
lista = [ 1 , 3, 5, 6, 2]

filtrado = filter(lambda x: x >= 5, lista)

print(list(filtrado))
````

Or implementing your own function:

````python
lista = [ 1 , 3, 5, 6, 2]

def mas_cinco(x):
    if x >= 5:
        return True
    
    return False

filtrado = filter(mas_cinco, lista)

print(list(filtrado))
````

As you can see, these are faster and more elegant ways to apply operations on collections.

## 6. Timeit
For the bootcamp, and generally when doing descriptive data analysis, runtime may not be critical. Imagine you created a heavy classifier that takes 0.5 seconds per transaction. If a user has 1000 transactions, the total time becomes 250 seconds (~4 minutes). Functional programming techniques can greatly improve performance because they avoid certain types of iteration.

Therefore, **when working with large datasets, consider using functions like `map`, `reduce`, and `filter` because they can yield significant performance improvements.**

````python
num_elementos = 100000

lista = list(range(num_elementos))
````

````python
%%timeit

lista_output = []
for i in lista:
    lista_output.append(i + 5)
````

````python
%%timeit

lista_output = map(lambda x: x + 5, lista)
````

As you can see, we can go from milliseconds to faster operations.

## 7. Functional programming in Pandas
Functional programming is useful when applying operations to every element of a column. In pandas we use the `apply` method.

Suppose we have a DataFrame of temperatures:

````python
import pandas as pd

weather = pd.DataFrame({"temperatura": [20, 15, 34, 4, 1, 25, 21, 29, 40]})
weather
````

If we want to compute a new variable that indicates whether it's summer or winter, we can apply a custom function via `apply`. First we define the function.

````python
def season(temp):
    if temp < 5:
        return "Winter"
    elif temp > 30:
        return "Summer"
    else:
        return "Other"
    
weather["season"] = weather["temperatura"].apply(season)
weather.head()
````

You can also use a lambda function.

````python
weather["more_temperature"] = weather["temperatura"].apply(lambda x: x + 5)
weather.head()
````

````python
weather["season_upper"] = weather["season"].apply(str.upper)
weather.head()
````

````python
# (Degrees °C × 9 / 5) + 32 = F

weather["temperatura_F"] = weather["temperatura"].apply(lambda x: (x *9/5)+32)
weather
````

````python
import numpy as np

weather['Business'] = np.where(weather['temperatura']>=20, True, False)
weather.head()
````
