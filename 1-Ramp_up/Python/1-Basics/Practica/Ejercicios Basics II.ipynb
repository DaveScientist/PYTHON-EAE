## Exercise 2
Write a program to convert degrees to radians. You must use `input`. Remember that the conversion is done through

radians = degrees*(pi/180)

````python
import math
from math import pi

degree = float(input("Insert the degrees: "))
radian = degree*(pi/180)
print(radian)
````

## Exercise 7
Create a program that takes two inputs. It must check if their sum is equal, greater than, or less than 10

````python
inp_1 = float(input("Insert the first number: "))
inp_2 = float(input("Insert the second number: "))

add = inp_1 + inp_2

print("Is greater than 10: ", add > 10)
print("Is less than 10: ", add < 10)
print("Is equal to 10: ", add == 10)
````
