## 6. Data Types
Python is a strongly typed language. This means that the variables we use belong to a data type: integer number (int), real (float), text (String), or another type of object.

**Why is it important to know well what types the data is?** Because each data type has a series of properties and operations associated with it. For example, you cannot add 5 to text. So when we go to do operations between them, we have to make sure they are the same type so that the result is what we expect. `text + 5` doesn't make sense and will give an error. It seems obvious, but there are times when the data types are not what we expect.

**How many data types are there?** Basically there are no limits here. In this notebook we will see the most basic ones, but in later notebooks you will see that you can create your own data types through **classes**. In Python there are a series of basic data types, which we can use without importing any external module, they are called [*built-in Types*](https://docs.python.org/3/library/stdtypes.html). These are the most common:

* **Numeric**: we have `int`, `float` and `complex`. Depending on whether it is an integer number, a real one or a complex one.
* **String**: or String. Plain text string
* **Boolean**: or Logical. `True`/`False`

**How do we know the data type of a variable?** Using `type(variable_name)`

## 7. Conversión de tipos
Como en otros lenguajes, en Python también tenemos la posibilidad de realizar conversiones entre los tipos de datos. Hasta ahora hemos visto pocos, pero cuando descubramos más tipos de datos y objetos, verás que son muy comunes estas transformaciones.

Un caso muy habitual es leer un archivo datos numéricos, y que Python interprete los números como caracteres. No es un error, pero posteriormente, cuando hagamos operaciones con nuestros números, tendremos errores, ya que en realidad son cadenas de texto. Si forzamos el cambio a numerico, nos evitaremos futuros problemas.

**Mucho cuidado en los cambios de tipos**. Tenemos que estar seguros de lo que hacemos ya que podemos perder información, o lo más probable, puede dar error, al no ser compatible el cambio.

Veamos como cambiar los tipos

```python
num_real = 24.69
print(type(num_real))

num_entero = int(num_real)
print(num_entero)
print(type(num_entero))
```

We lose some decimals, usually nothing serious, although it depends on the application.

**NOTE**: if what we want is to round, we can use the `round()` function

```python
round(24.691739372, 0)
```

```python
type(int(round(24.691739372, 0)))
```

```python
round(num_real, 1)
```

## 7. Type Conversion
As in other languages, in Python we also have the possibility to perform conversions between data types. So far we have seen few, but when we discover more data types and objects, you will see that these transformations are very common.

A very common case is reading a file with numerical data, and Python interpreting the numbers as characters. It's not an error, but later, when we do operations with our numbers, we'll have errors, since they are actually text strings. If we force the change to numeric, we'll avoid future problems.

**Be very careful when changing types**. We have to be sure of what we're doing as we can lose information, or more likely, it can give an error, as the change may not be compatible.

Let's see how to change the types

```python
numero = "98,25"
print(numero)

numero_punto = numero.replace(",", ".")
print(numero_punto)

numero_float = float(numero_punto)
print(type(numero_float))
```

```python
string_ejemplo = "I live in Madrid. Sure in Madrid?"

print(string_ejemplo.replace("Madrid", "Barcelona"))
```

It is fundamental to operate with the same data types. Look what happens when we add text with a number. It gives a `TypeError`. Basically it tells us that you can't concatenate a text with an integer

```python
"4" + 6
```

```python
6 + "4"
```

## 10. Syntax and Best Practices
When writing in Python, there are certain rules you need to follow:
* Everything you open, you have to close: parentheses, braces, brackets...
* Decimals are placed with dots `.`
* Best practices
    * **Characters**: It is NOT recommended to use Ñs, accents or strange characters (ª,º,@,ç...) in the code. Put it only in the comments.
    * **Spaces**: Do NOT use spaces in the names of variables or functions. It is recommended to use an underscore to simulate the space. Or also put the words together and use a capital letter to differentiate them `myVariable`. The norm is all lowercase and underscores
    * However, it is recommended to use spaces between each command, to facilitate reading, although this is more a matter of taste. `my_variable = 36`.
    * Variables are usually declared in lowercase.
    * Constants (variables that will never change) in uppercase. `MY_COUNTRY = "Spain"`
    * **Each statement on a line**. You can use `;` to declare multiple variables, but it's not the norm
    * **Comments**: As many as possible. You never know when someone else is going to pick up your spectacular code, or if your *future self* will remember why you did that while loop instead of a for.
    * **Case sensitive**: sensitive to uppercase and lowercase letters. BE CAREFUL with this when declaring variables or using Strings
    * **Line syntax**: for proper code reading, it's best to apply line syntax as much as possible
