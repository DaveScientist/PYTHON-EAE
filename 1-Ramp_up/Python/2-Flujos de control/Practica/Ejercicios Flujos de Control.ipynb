## Exercise 1
Declare a numeric variable that will be any hour of the day. Implement the following cases using `if/elif/else` statements:

* If it is between 0 and 8, print "Sleeping"
* If it is between 9 and 18, print "Working"
* If it is between 19 and 21, print "Class"
* If it is between 22 and 24, print "Rest"
* In any other case, print "Transport or error"

```python
hour = 24

if hour >= 0 and hour <= 8:
    print("Sleeping")
elif hour >= 0 and hour <= 18:
    print("Working")
elif hour >= 0 and hour <= 21:
    print("Class")
elif hour >= 0 and hour <= 24:
    print("Rest")
else: 
    print("Transport or error")

```

## Exercise 2
In this exercise we are going to implement a very simple house price calculator. We have the following variables:

> surface
>
> district

Implement using `if/elif/else` statements the following cases:

1. If the district is "Moncloa" or "Centro", and the surface area is greater than 100 square meters, the price of the house is 1000
2. If the district is "Salamanca", and the surface area of the house is at least 150 meters, the price of the house is 1500
3. If the district is not "Retiro" and the surface area is between 60 and 80 meters, the price is 600
4. In any other case, the price will be 0

```python
surface = 100
district = "Moncloa"

if district in ("Moncloa", "Centro") and surface == 100:
    print("The price of the house is 1000")
    
elif district in ("Salamanca",) and surface == 150:
    print("The price of the house is 1500")
    
elif district not in ("Retiro",) and 60 <= surface <= 80:
    print("The price of the house is 600")
    
else:
    print("Price not found, the values are not valid")
```

## Exercise 8
Implement a program that prints on screen the following pattern

5 4 3 2 1 

4 3 2 1 

3 2 1 

2 1 

1

NOTE: There are NO blank lines between one line and another.

```python
for i in range(5, 0, -1):
    for j in range(i, 0, -1):
        print(j, end=" ")
    print()
```

## Exercise 9
In this exercise we are going to create a small game. It is about trying to guess a number from 1 to 5. We have two attempts to get it right. Steps to follow:

1. How to get a random number from 1 to 5 is already implemented
2. You will have to declare in a variable the number of lives, and through a while loop, check that there are still lives left.
3. Inside the loop, get the value from the user and check if that is the number to guess. If not, update the lives.
4. If we guess correctly, we exit the loop and print "You win" on screen. And if we lose, we also exit the loop, but in this case we print "You lose" on screen.

TIP: It might be useful to use the `else` statement when the `while` loop ends. What is inside that `else` will be executed once the `while` execution ends. You can use it for when you lose.

```python
import random

secret_number = random.randint(1, 5)

lives = 2

while lives > 0:
    guess = int(input("Guess a number from 1 to 5: "))

    if guess == secret_number:
        print("You win")
        break
    else:
        lives -= 1
        print(f"Wrong! Lives left: {lives}")
        
else:
    print("You lose")
```
