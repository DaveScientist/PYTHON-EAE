## Exercise 1
Declare a numeric variable that will be any hour of the day. Implement the following cases using `if/elif/else` statements:

* If it is between 0 and 8, print "Sleeping"
* If it is between 9 and 18, print "Working"
* If it is between 19 and 21, print "Class"
* If it is between 22 and 24, print "Rest"
* In any other case, print "Transport or error"

```python
hour_of_day = int(input("Tell me the hour: "))

if hour_of_day >= 0 and hour_of_day <= 8:
    print("Sleeping")
elif hour_of_day >= 9 and hour_of_day <= 18:
    print("Working")
elif hour_of_day >= 19 and hour_of_day <= 21:
    print("Class")
if hour_of_day >= 22 and hour_of_day <= 24:
    print("Rest")
else:
    print("Error")
```

## Exercise 6
Given the following list:

```python
list1 = [12, 15, 32, 42, 55, 75, 122, 132, 150, 180, 200]
for i in list1:
    if i % 5 == 0:
        print(i)
    if i > 150:
        break
```

## Exercise 9
In this exercise we are going to create a small game. It is about trying to guess a number from 1 to 5. We have two attempts to get it right. Steps to follow:

1. How to get a random number from 1 to 5 is already implemented
2. You will have to declare in a variable the number of lives, and through a while loop, check that there are still lives left.
3. Inside the loop, get the value from the user and check if that is the number to guess. If not, update the lives.
4. If we guess correctly, we exit the loop and print "You win" on screen. And if we lose, we also exit the loop, but in this case we print "You lose" on screen.

TIP: It might be useful to use the `else` statement when the `while` loop ends. What is inside that `else` will be executed once the `while` execution ends. You can use it for when you lose.

```python
import random
random.randint(1,5)
```

```python
random_number = random.randint(1,5)
#print(random_number)
lives = 2

while lives > 0:
    guess = int(input("Guess the number:"))
    if guess == random_number:
        print("You win")
        break
    lives -= 1
else:
    print("You lose")
```
