## 1. Line Syntax
The way Python encapsulates all the code that goes inside a control flow like `if` or `for` is different from how it's usually done in other languages, where it's surrounded by braces `{}` or parentheses `()`. Not in Python. In Python you just have to add an indentation to each line of code that goes inside the control flow.

> ```Python
> for conditions:
>     Code inside this loop
> ```


If we leave it out, this code will execute sequentially after the for runs

> ```Python
> for conditions:
>
> Code outside this loop
> ```

Let's see an example. We have a list of numbers, and we want to see which ones are integers. To do this we iterate through them with a `for` (we will see more in depth in this notebook). We iterate one by one each element. Then through an `if` we check if it is an integer. Notice that everything inside the `for` has one indentation and what is inside the `if` has two indentations, since its statements go both inside the `if` and inside the `for`.

````python
numbers = [4, 6, 4.0, 3.0]

for element in numbers:

    if type(element) == int:
        print("The number", element, "is an integer")
````

## 2. if/elif/else
Depending on what certain boolean conditions are worth, we will execute some lines of code, or others. The syntax is very simple:


> ```Python
> if conditions:
>     If the conditions are met, execute this code
> else:
>     If not, execute this other code
> ```

Let's see an example

Only the part of the code that gets a `True` in the condition is executed. `print("Passed!")` will only be printed to the screen if the grade is 5 or higher.

Let's look at another example. Now I want a bit more granularity in the grade, with satisfactory, good, etc.

````python
my_grade = 10

if my_grade < 5:
    print("Summer school")
    # if type(my_grade) == int:
    #     print("It's an integer")
    # if type(my_grade) == float:
    #     print("It's a float")
    
elif my_grade < 6:
    print("Satisfactory")
    
elif my_grade < 7:
    print("Good")
    
elif my_grade < 9:
    print("Very good")
    
else:
    print("Excellent")
````

**IMPORTANT**. All the `ifs` are executed sequentially. That's why, in this case, it's not necessary to restrict the grade so much:

* First it checks if it's less than 5, if so, fail.
* We already know it's 5 or higher
* In the next condition it checks if it's less than 6, that is, between 5 and 6. If it's `False`, we move to the next condition. However, if it's `True`, we execute only that code and forget everything else. There could be 150 conditions, but if the first one is `True`, the rest is as if it doesn't exist.

Note that the syntax is quite intuitive. `if` a condition, colon and execute everything inside here, `elif` (short for `else if`), if this other condition is met, execute this other code. And if none of these are met, execute what's in the `else`.

````python
beak = False
wings = False
sound = "Barking"
legs = 4

if beak or wings:
    print("Bird")

elif legs == 4 and sound == "Barking":
    print("Dog")

else:
    print("Other animal")
````

## 3. For Loop
Thanks to loops we can execute repetitive code, in a quite automatic way. They are very useful so that our code is not redundant, and also to apply operations when we handle iterables. An iterable is simply a collection of objects (a list is an iterable) that we can iterate through one by one with the `for` loop, and apply operations to each element.

The syntax of the `for` loops is as follows:

> `for var_execution in execution_limits:`
> 
>           code of the for...
    
    
* **Execution limits**: The number of times we want a `for` to execute. This is because otherwise it would execute until infinity. And also, they have an execution variable that is updated. For example from 1 to 10. First it would be worth 1, then 2...until 10.


* **Execution variable**: inside the for there will be a variable that will be updated with each execution. If it executes 10 times, first the variable will be worth 1, then 2, and so on until 10.

![imagen](./img/for-loop-python.jpg)

Better let's see an example to understand it. You have the grades of three students in a list, and you want to print the grades to the screen

````python
grades = [3, 6, 9, 10]

print(grades[0])
print(grades[1])
print(grades[2])
print(grades[-1])
````

````python
for grade in class_grades:
    if grade >= 5:
        print("Passed with:", grade)
    else:
        print("Failed with:", grade)
````

## 4. While Loop
This is another way to implement a loop in programming. Loops always have to be limited. In the case of the `for`, we put a specific number of executions, depending on the iterable we were iterating through. For `while` it's something different. It has an **execution condition**, that as long as it is met (`True`), it will keep executing over and over again. On the other hand, the loop has an **execution variable**, just like in the `for`, which will be updated with each iteration, and it's that variable that determines when the loop ends.

![imagen](./img/while-loop.png)

**Be careful** with these loops as it's very easy to forget to update the execution variable, or make a mistake in the execution condition. If this happens the code will keep running until we stop the kernel (button *interrupt the kernel*, above next to the Run)

Let's see an example.

````python
i = 0

while i < 5:
    print(i)
    
    i = i + 1 # i += 1
    
print("End program")
````

````python
i = 0
import time

while i < 5:
    print(i)
    time.sleep(2)
    
print("End program")
````

The most common way to implement these loops is:
1. I declare the **execution variable outside the loop**
2. I establish an **execution condition** to determine when we want the loop to stop.
3. I **update the execution variable** in each iteration of the loop.

## 5. Break/continue
These are two statements that we can use inside loops to avoid executing unnecessary code.

### Break
It's used when we want to forcefully exit the loop. Imagine you're a store and you're searching with a for if at least one of the orders was a coat. If you had 1000 orders, you would have to iterate over all of them and through an `if`, check if it's a coat. Now, if the coat is the first element in the list, the `for` will still iterate over the other 999 elements, when it's not necessary. With a `break` we can exit the loop and continue with the program.

![imagen](./img/break.jpg)

````python
for i in [1,2,3,4]:
    print(i)
    if i > 1:
        break
````

````python
for val in "string":
    if val == "i":
        break
        
    print(val)

print("End")
````

### Continue
This statement is used inside a loop to tell it to continue with the next element of the iterable. Like with `break`, it helps us avoid executing unnecessary code. Going back to the previous example, after checking that we have a coat, there are 200 more lines of code that are used in other cases, with a `continue` we avoid executing all of that, we do what we have to do with the coat, and we tell the loop to move to the next element, and ignore the rest of the code.

![imagen](./img/continue.jpg)

````python
for val in "string":
    if val == "i":
        continue
        # print("hello")
        
    print(val)


print("End")
````

**The `for` and `while` loops, as well as `break` and `continue`, are complicated statements to understand, and if this is your first time programming it will require a change in the way you think and solve problems, so I recommend that you get a pen and paper and do the first loop exercises by seeing the iterations one by one and manually calculating all the operations inside the loop.**

## 6. Try/except
What happens when there's an error in our code? The entire execution stops. However good programmers we are, we always have to account for the fact that there can be errors. We can manage to control them with `if/else` statements, for example if we don't know much about the types of data, `if type(data) == float:` you do something with floats, `else` you do something else with another type of data, but the best thing is to use `try/except`.

Now, if we suspect that the behavior of our code can be somewhat unpredictable, in programming we can use `try/except` statements to catch that error, make decisions, and have the code continue executing.

The syntax is as follows:

> ```Python
> try:
>     Code that may contain errors
> except:
>     What to do if we encounter an error
> ```

````python
print(variable_)
print("AAA")
````

````python
try:
    print(variable_)
except NameError:
    print("The code has errors")
    pass
    
finally:
    print("It executes anyway")
    
print("AAA")
````

````python
try:
    print(variable_)
except:
    print("The code has errors")
    print("It executes anyway")
    pass
    
print("AAA")
````

````python
list_ = [1,2,3]
print(list_[5])
print("Hello world")
````

````python
list_ = [1,2,3]

try:
    print(list_[5])

except IndexError:
    print("Error in the name")
    
except:
    print("Another error")
````

````python
list_ = [1,2,3]

try:
    print(list_[5])
except Exception as e: print(e)
print('AAA')
try:
    print(variable_)
except Exception as e: print(e)
print('BBB')
````

There's an error in the code, but it doesn't stop the program.

We can be a bit more specific with the errors, and depending on the type of error we get, we can take different paths
