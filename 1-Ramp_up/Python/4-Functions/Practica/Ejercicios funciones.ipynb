## Exercise 1
Write a function that converts numbers from 1 to 7 into names of the days of the week. The function will consist of a single numeric argument and a string type output

````python
def dias_semana(nombre):
    if nombre == "Lunes":
        return 1
    elif nombre == "Martes":
        return 2
    elif nombre == "Miércoles":
        return 3
    elif nombre == "Jueves":
        return 4
    elif nombre == "Viernes":
        return 5
    elif nombre == "Sábado":
        return 6
    elif nombre == "Domingo":
        return 7
    else:
        return "Nombre de día no válido"
    
dias_semana("Domingo")
````

In exercise 8 of loops, we created an inverted pyramid, whose number of floors was determined by user input. Create a function that replicates the behavior of the pyramid, and uses a single input parameter of the function to determine the number of rows of the pyramid, that is, removes the input statement.

````python
def pyramide_invertida(n): 
   for i in range(n, 0, -1):          # 5 → 1
      for j in range(i, 0, -1):      # i → 1
         print(j, end=" ")
      print()

pyramide_invertida(8)           
````

## Exercise 3
Write a function that compares two numbers. The function has two arguments and there are three possible outputs: they are equal, the first is greater than the second, or the second is greater than the first

````python
def compare(num1, num2):
    if num1 > num2:
        return "num1 es mayor que num2"
    elif num1 < num2:
        return "num1 es menor que num2"
    else:
        return "num1 y num2 son iguales"
compare(5, 10)
````

## Exercise 4
Write a function that is a letter counter. In the first argument you have to introduce a text, and the second that is the letter to count. The function must return an integer with the number of times that letter appears, both uppercase and lowercase

````python
#Preguntar el rol del count
def letter_counter(string, letter):
    count = 0           
    for char in string:
        if char == letter:
            count += 1
    return count
letter_counter("banana", "a")
````

## Exercise 5
Write a function that has a single argument, a string. The output of the function must be a dictionary with the count of all the letters of that string.

````python
def dictionary(letter):
    dict = {}
    for char in letter:
        if char in dict:
            dict[char] += 1
        else:
            dict[char] = 1
    return dict
dictionary("banana")
````

## Exercise 6
Write a function that adds or removes elements in a list. The function needs the following arguments:
* list: the list where elements will be added or removed
* command: "add" or "remove"
* element: By default it is None.

````python
def elements(lista, command, elemento = "None"):
    if command == "add":
        lista.append(elemento)
        return lista
    elif command == "remove":
        if elemento in lista:
            lista.remove(elemento)
            return lista
        else:
            return "Elemento no encontrado en la lista"
    else:
        return "Comando no válido"
print(elements([1, 2, 3], "add", 4))
print(elements([1, 2, 3], "remove", 2))
````

## Exercise 7
Create a function that receives an arbitrary number of words, and returns a complete sentence, separating the words with spaces.

````python
def arbitrary(words):
    frase = ""
    for i in words:
        frase += i + " "
    print(frase.strip())

arbitrary(["Hola", "mundo", "Python"])
````

## Exercise 8
Write a program that obtains the nth number of the [Fibonacci sequence](https://en.wikipedia.org/wiki/Fibonacci_number). You must create a recursive function with a single argument.

```Python
fibonacci(2) = 1
fibonacci(4) = 3
fibonacci(5) = 5
fibonacci(8) = 21
```

````python
#Preguntar!
def fibonacci(n):
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci(n-1) + fibonacci(n-2)
    
fibonacci(8)
````

## Exercise 9
Define in a single cell the following functions:
* Function that calculates the area of a square
* Function that calculates the area of a triangle
* Function that calculates the area of a circle

In another cell, calculate the area of:
* Two circles with radius 10 + a triangle with base 3 and height 7
* A square with side = 10 + 3 circles (one with radius = 4 and the other two with radius = 6) + 5 triangles with base = 2 + height = 4

````python
#Dictionary puede ser con =??
# en este caso para que sirve get?

def calculate_area(shape, **kwargs):
    if shape == "circle":
        radius = kwargs.get("radius")
        if radius is not None:
            return 3.14159 * radius ** 2
        else:
            return "Falta el radio para calcular el área del círculo"
    elif shape == "square":
        side = kwargs.get("side")
        if side is not None:
            return side ** 2
        else:
            return "Faltan el ancho o la altura para calcular el área del rectángulo"
    elif shape == "triangle":
        base = kwargs.get("base")
        height = kwargs.get("height")
        if base is not None and height is not None:
            return 0.5 * base * height
        else:
            return "Faltan la base o la altura para calcular el área del triángulo"
    else:
        return "Forma no reconocida"

print(calculate_area("circle", radius=5))
print(calculate_area("square", side=4))
````
