## Exercise 1
Write a function that converts numbers from 1 to 7 into names of the days of the week. The function will consist of a single numeric argument and a string type output

````python
def weekday(day_number):
    if day_number == 1:
        return "Monday"
    elif day_number == 2:
        return "Tuesday"
    elif day_number == 3:
        return "Wednesday"
    elif day_number == 4:
        return "Thursday"
    elif day_number == 5:
        return "Friday"
    elif day_number == 6:
        return "Saturday"
    elif day_number == 7:
        return "Sunday"
    else:
        return "Invalid day number"
````

````python
weekday(1)
````

## Exercise 2
In exercise 8 of loops, we created an inverted pyramid, whose number of floors was determined by user input. Create a function that replicates the behavior of the pyramid, and uses a single input parameter of the function to determine the number of rows of the pyramid, that is, removes the input statement.

````python
def pyramid(rows):
    
    while rows > 0:
        for i in range(rows, 0, -1):
            print(i, end=" ")
        print()
        rows = rows - 1
    return "Pyramid completed"
    
````

````python
pyramid(5)
````

## Exercise 3
Write a function that compares two numbers. The function has two arguments and there are three possible outputs: they are equal, the first is greater than the second, or the second is greater than the first

````python
def compare(num1, num2):
    num1 = float(num1)
    num2 = float(num2)
    if num1 == num2:
        return "The numbers are equal"
    elif num1 > num2:
        return str(num1) + " is greater than " + str(num2)
    else:
        return str(num2) + " is greater than " + str(num1)
````

````python
compare(5, 10)
````

## Exercise 4
Write a function that is a letter counter. In the first argument you have to introduce a text, and the second that is the letter to count. The function must return an integer with the number of times that letter appears, both uppercase and lowercase

````python
def counter_letters(text, letter):
    text = text.lower()
    return text.count(letter.lower())
````

````python
counter_letters("This class is awesome", "A")
````

## Exercise 5
Write a function that has a single argument, a string. The output of the function must be a dictionary with the count of all the letters of that string.

````python
def letter_count_dict(text):
    all_letters = {}

    for i in text.lower():
        if i == ' ':
            continue
        if i in all_letters:
            all_letters[i] = all_letters[i] + 1
        else:
            all_letters[i] = 1
            
    return all_letters
````

````python
def letter_count_dict_2(text):
    all_letters = {}
    text = text.lower()

    for i in text:
        if i not in all_letters:
            all_letters[i] = text.count(i)
        else:
            continue

    return all_letters
````

````python
letter_count_dict("Python is great")
````

## Exercise 6
Write a function that adds or removes elements in a list. The function needs the following arguments:
* list: the list where elements will be added or removed
* command: "add" or "remove"
* element: By default it is None.

````python
def exercise_6(lista, command, element = None):
    lista = list(lista)

    if command == "add" and element != None:
        lista.append(element)
        return lista
    
    elif command == "add" and element == None:
        # print("You didn´t provide an element to add")
        return lista

    elif command == "remove" and element != None:
        try:
            lista.remove(element)
        except:
            print("The element", element, "is not in the list")
        print(lista)

    elif command == "remove" and element == None:
        print("You didn´t provide an element to remove")

    else:
        print("Invalid command")

````

````python
print(exercise_6((1, 2, 3), "add", 7))
````

## Exercise 8
Write a program that obtains the nth number of the [Fibonacci sequence](https://en.wikipedia.org/wiki/Fibonacci_number). You must create a recursive function with a single argument.

```Python
fibonacci(2) = 1
fibonacci(4) = 3
fibonacci(5) = 5
fibonacci(8) = 21
```

````python
def fibonacci(n):
    if n == 1:
        return 1
    elif n == 0:
        return 0
    else:
        return (fibonacci(n-1) + fibonacci(n-2))
````

````python
print(fibonacci(2))
print(fibonacci(10))
print(fibonacci(5))
print(fibonacci(3))
````
