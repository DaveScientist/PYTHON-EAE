## Exercise 1
Write a function that converts numbers from 1 to 7 into names of the days of the week. The function will consist of a single numeric argument and a string type output

````python
def weekday(day_number):
    if day_number == 1:
        return "Monday"
    elif day_number == 2:
        return "Tuesday"
    elif day_number == 3:
        return "Wednesday"
    elif day_number == 4:
        return "Thursday"
    elif day_number == 5:
        return "Friday"
    elif day_number == 6:
        return "Saturday"
    elif day_number == 7:
        return "Sunday"
    else:
        return "Invalid day number"
````

````python
weekday(500)
````
