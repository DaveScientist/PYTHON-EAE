## 1. Definition, syntax and return
Through **functions** we can encapsulate code in input/output format. So if you have repetitive code that depends on certain inputs, functions can be a good solution.

![imagen](./img/funciones.png)

It is a way to group sets of operations into modules. **When to use them?** When we have several operations that we execute repeatedly in different parts of the code. In that case, we encapsulate the operations in a function, and each time we need to perform that operation, we call the function, and in a single line of code we have executed those operations.

So far we have been using *built-in* functions, for simple operations like `len()`, `sum()` or `max()`. In this Notebook you will learn to create your own functions.

The syntax is:
> ```Python
> def function_name(input):
>    various operations
>    return output
> ```

Note that it follows the **line syntax** seen in previous Notebooks. Also, everything after `return` is ignored, since it is the output. At `return` the function ends. However, that doesn't mean there's only one return. If we introduce an `if/else` statement, we can put different returns depending on which condition is met. Let's create our first function

````python
# Let's create a km to miles converter
def km_millas(distancia_km):
    distancia_millas = distancia_km * 0.62
    return distancia_millas
````

````python
millas = km_millas(10)
print(millas)
````

````python
type(km_millas)
````

When executing the code above, nothing runs, we simply store the function in memory to use it later. **Now we will be able to call the function as many times as we want, from any part of the code.**

````python
print(km_millas(2))
print(km_millas(5))
print(km_millas(10))
````

Functions don't necessarily have to carry arguments. However, **it is mandatory** to put parentheses, both in the declaration and then when calling the function.

````python
from datetime import datetime

def que_hora_es():
    now = datetime.now().time()

    print(now)
    return now
````

````python
hora = que_hora_es()
````

````python
def que_hora_es():
    time = datetime.now().time()
    date = datetime.now().date()
    return time, date

print(que_hora_es())
````

## 2. Positional Arguments
You already know how to create functions with a single argument. You will have the option to implement them with as many arguments as you want. However, keep in mind two things:

1. **The order** of the arguments. When we call the function, we have to follow the same order of arguments as in the function declaration.
2. **They are mandatory**. If we declare them in the function, then when calling it, we have to put all its arguments. Later we will see that there is a way to put optional arguments.

````python
def multiplica(x1, x2, x3, x4):
    return (x1*x2*x3)/x4

multiplica(4,6,7,2)
````

Notice that the arguments follow a certain order: x1, x2, x3, x4. When we call the function, we will introduce 4 arguments and the function will collect them in that order. It will assign 4 to x1, 6 to x2, etc. We can also specify the argument name in the call, which allows us to have more flexibility in the order.

````python
multiplica(x4 = 2,
           x2 = 6,
           x3 = 7,
           x1 = 4)
````

## 3. Variable Arguments
In the previous examples we had to fix a specific number of arguments, but sometimes we're not sure how many arguments there are. Fortunately, Python functions provide us with that flexibility through `*`

Let's see how to implement a multiplier function with a variable number of arguments

````python
def multipl_var(*args):
    #print(type(args))
    #print(args)
    
    total = 1
    for i in args:
        total = total * i
        
    return total
    
    
print(multipl_var(2, 5))
print(multipl_var(2, 5, 6, 7))
print(multipl_var(2))
````

Keep in mind that `*args` is a variable with X elements. Since we don't know a priori how many there are, we will have to iterate through them with a `for`, and for each argument, apply an operation. Therefore, `*args` is an iterable, specifically a **tuple**. What gives it the functionality of "variable arguments" is `*`, not `args`. Just as we put `*args`, we can put `*arguments`.

You can combine positional arguments with `*args`

````python
# In this example, I use the last argument to divide everything we had multiplied by this argument
def multipl_var_div(*args, div):   
    total = 1
    for i in args:
        total = total * i
    return total/div

multipl_var_div(2,5,3,div = 6)
````

````python
def multipl_var_div(div, *args):   
    total = 1
    for i in args:
        total = total * i
    return total/div

multipl_var_div(6,2,5,3)
````

### Variable arguments with key-value
We also have the option to introduce a dictionary as arguments, so although the number of arguments is variable, we will have an indicator, the key, and the value of each key. It is implemented with `**`

````python
def movil(**args):
    print(args)
    print(type(args))
    
    for key, value in args.items():
        print(key, "=", value)
    
movil(Camera = "24MPx", Battery = 10, Weight = 200)
````

### Combining `*args` with `**kwargs`
There is no problem in having a variable number of arguments and also key-value arguments, all in the same function.

````python
def movil(*args, **kargs):
    
    for i in args:
        print(i)
        
    for key, value in kargs.items():
        print(key, "=", value)
    
movil("Good mobile",
      "Battery lasts a long time",
      Camera = "24MPx",
      Battery = 10,
      Weight = 200)
````

## 4. Keyword Arguments
There is another type of arguments that are *keyword*. They are characterized because they have a default value, and therefore, if we don't use that argument in the call, inside the function it will take the value we left by default.

Keep in mind that these arguments are placed **at the end**

````python
def venta_online(order, delivery_date, incident=False):
    
    if incident:
        print("Contact Customer Service")
    
    else:
        print("Your order", order, "will be delivered on", delivery_date)

venta_online("AAA", "18-07-2021")
venta_online("AAA", "18-07-2021", False)
venta_online("AAA", "18-07-2021", True)
````

## 5. Recursion
A function can call itself in its own declaration, as if it were a loop. It is a somewhat complex concept, but elegant when it comes to implementing our programs. The only negative part is that it's a bit difficult to understand what the function does. Like everything, it has its advantages and disadvantages.

Let's calculate the factorial of a number *!n*

````python
list(range(3))
````

````python
# We could calculate it with a loop
num_factorial = 6

output = 1

for i in range(num_factorial):
    output = output*(i+1)
    print(i+1)
    
print('\n',output)
````

````python
# Or through a recursive function
def factorial(x):
    
    if x == 1:
        return 1
    
    else:
        return x * factorial(x-1)
````

## 6. Documenting functions
As we saw in the first Notebook, you have to document the code as much as possible. In particular, it is necessary to document functions well, because many times we import them from elsewhere, we use them because they work, but we don't really know what they do. That is why in Python there is an attribute within functions, modules, methods or classes, that allows access to "their comments", their documentation, which tells us what it does.

This special attribute is called *docstring*, and is accessed using `function_name.__doc__`

````python
def multiplica(x, y):
    '''
    Function that multiplies the two arguments: x*y
    Inputs:
        x: float
        y: float
        
    Output:
        x * y: float
    '''
    return x*y

print(multiplica(2,2))
print(multiplica.__doc__)
````

The comments that are added can be single-line or multi-line. For simple functions, a single line comment may be sufficient, but if they were more complex, the *docstring* should carry the following information:
* Description of the function
* Input arguments: name, types and what they do
* Output arguments: name, types and what they are
