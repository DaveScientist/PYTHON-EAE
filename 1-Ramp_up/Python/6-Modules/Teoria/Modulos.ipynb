## 1. import
Remember that you should **avoid having repeated or duplicate code as much as possible**, so functionalities that you use in different parts of the code you don't have to copy and paste everywhere, but there is the possibility to define them once, and import them later.

Therefore, as we complicate our programs, the need arises to modularize them, that is, to be able to divide them and package them. As you may have imagined, productive programs like a website or a game, don't go in a Notebook, but in several Python scripts. According to the different functionalities of the code, we will divide it into several parts. For example, auxiliary functions can go in one script, constants in another, your Classes in another... Let's see some concepts

**Script**

It's just a file with Python code. These files have the `.py` extension, so an example could be `first_script.py`.

**Module/Library/Package**

These are similar terms that are usually used to refer to the same thing. For practical purposes, we will consider them equal. These are packages of code that have a well-defined functionality and we can import in our Python programs, using what is inside (classes, functions, variables...).

**import**

Reserved word and statement that is used in Python to import any object in our code. It is used to import variables, functions or modules.

Let's see how we can import objects from other scripts

```python
%%file first_script.py
"""
Example of a python module. Contains a variable called my_variable,
a function called my_function, and a class called MyClass.
"""

my_variable = 0

def my_function():
    """
    Example function
    """
    return my_variable
    
class MyClass:
    """
    Example class.
    """

    def __init__(self):
        self.variable = my_variable
        
    def set_variable(self, new_value):
        """
        Set self.variable to a new value
        """
        self.variable = new_value
        
    def get_variable(self):
        return self.variable
```

We have in another script a variable, a function and a class defined. Now let's import it

```python
import first_script
help(first_script)
```

As always in any programming language **we have to always pay attention to where the Notebook is pointing**, that is, in what working directory it will read the script. When you do `import first_script`, it will look for that script in the **same directory where the Notebook is**. If it's somewhere else on the computer, it won't detect it. When you do the import, Python doesn't search through all the folders on the computer, it looks for it in the same place where the Notebook is. Later we will see how to read scripts from other directories.

If we want to access any object from the module, we simply use the syntax `module_name.object`

```python
"""
Example of a python module. Contains a variable called my_variable,
a function called my_function, and a class called MyClass.
"""

my_variable = 0

def my_function():
    """
    Example function
    """
    return my_variable
    
class MyClass:
    """
    Example class.
    """

    def __init__(self):
        self.variable = my_variable
        
    def set_variable(self, new_value):
        """
        Set self.variable to a new value
        """
        self.variable = new_value
        
    def get_variable(self):
        return self.variable
```

```python
print(first_script.my_variable)
print(first_script.my_function)
print(first_script.MyClass)
```

```python
new_object = first_script.MyClass()
print(new_object)
print(new_object.get_variable())
```

Finally, **what if our modules are not in a folder within the project we are working on, but in another folder on the computer, or in parent directories?** We can solve this using `sys.path`. `sys` is the Python module used to handle system variables and the interpreter. And `path` is a list of strings with a series of paths where the Python interpreter goes when it has to search for a module. [You have here the documentation of `sys`](https://docs.python.org/3/library/sys.html)

```python
import sys
sys.path
```

If we want the Python interpreter to understand other modules that **are not in the Notebook folder**, we will have to add that path to the list of paths using the syntax `sys.path.append(new_path)`. Remember that `sys.path` is a list, so you can apply what you already know about lists to it.

## 2. Libraries
So far, all the functionalities you needed in Python you have been developing with your own code. This is fine, but it's not always necessary. **There are many libraries that can do the work for you**, so before implementing something, ***don't reinvent the wheel***. Be smart and search if there's already something done on the web. This is what's good about working with an open source language, that there is a community with many developers and there is a large amount of code published for free. So, where can we get these magical codes?

* **[Python Standard Library](https://docs.python.org/3/library/index.html#the-python-standard-library)**: here you have much of what we have already seen (data types, the math module...) and much more (file reading, compression, access to databases...).
* **Third-party Libraries**: libraries that do not come in the default Python package. These libraries are implemented either by an individual or a company, and are open source. Anaconda is very oriented towards data science, so it has very interesting libraries that we will use during the course (pandas, numpy or sklearn). In addition to these libraries, you will be able to add all the ones you need to the Python interpreter

Let's see an example of importing a library

```python
import math
math.sqrt(36)
```

By importing `math`, we will be able to access all its functions using the syntax `math.function()`. [In the documentation](https://docs.python.org/3/library/math.html#math.log) you will see everything you can do with `math`. 

Let's see an example that helps us not write more code.

### Third-party Libraries

So, **how do you install a new package?** You will find most packages in [PyPI (Python Package Index)](https://pypi.org), which is the official repository of Python packages. There are packages from individuals (not companies, or communities) in PyPI. That doesn't mean you can't install them, but they are probably less tested than the "official" ones.

Let's see an example of how to install a new library. In this case, we will install **wget**, which is used to obtain files from the web.
