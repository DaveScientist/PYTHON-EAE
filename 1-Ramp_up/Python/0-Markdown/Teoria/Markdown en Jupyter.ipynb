### What is it?
A markup language that allows us to apply formatting to our texts through special characters. Very useful when we have to document something, write an article, or deliver a report. This language is designed for the web, but it's very common to use it in any type of text, regardless of its destination.

The good thing about it is that it's edited in **plain text and integrated into many tools**, such as Jupyter Notebook or RStudio.

### Markdown vs HTML
We have an old acquaintance in web programming: HTML. They are very different languages. With HTML we can build a complex **tree of tags**, while markdown develops in plain text. Of course, the purposes are also different. HTML applies to all types of webs, whether simple or complex, while markdown is usually used for blogs or articles. Its simplicity in development penalizes it in its versatility. But since the objective of this course is not to build web pages, markdown is more than enough to accompany and improve the understanding of our code. Besides, as you will see throughout this notebook, both languages are perfectly compatible.

### How does it work?
It contains a series of **special characters** that give shape to texts. For example, if we want a text in *italics*, we simply surround it with asterisks. We will see this in detail in this Notebook.

## 4. Bold, Italics
To highlight text in bold we have to surround it with asterisks. In the case where we want italics, it will be a single asterisk, and if we want to combine bold with italics, it's 3 asterisks.


**Bold text**

*Italic text*

***Bold and italic text***

Be careful not to leave spaces between the asterisks and the text. That is, if we want to write in bold, the text must immediately follow the asterisks: ** This is not bold **

## 10. Images
If we have an image on the computer, we have to tell Markdown to point to that image. Images are NOT attached. The normal thing is to have all images grouped in a folder within your repository.

We use the syntax ![ any name](image path).

The image must be in the same folder as this notebook, due to the syntax `./image.png`. With the `./` Jupyter understands that it has to search in the folder of this notebook. If we put `../` we tell it that the image is in the previous directory:

1. `./image.png` if the image is in the same directory as this Notebook
2. `./img/image.png` if inside the directory where this Notebook is located, there is a folder called "img", and inside it is the image.
3. `../image.png` if the image we're looking for is in the directory above where this Notebook is located.
